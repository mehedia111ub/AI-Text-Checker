```python
# AI Text Detection using Hugging Face Transformers
from transformers import pipeline

# Load once (cached in Streamlit later)
def load_detector():
    classifier = pipeline(
        "text-classification",
        model="roberta-base-openai-detector",
        device=-1
    )
    return classifier

def detect_ai(text, classifier):
    result = classifier(text)[0]
    
    label = result['label']
    score = result['score']

    if "AI" in label.upper():
        ai_score = score
        human_score = 1 - score
    else:
        human_score = score
        ai_score = 1 - score

    return round(ai_score * 100, 2), round(human_score * 100, 2)
```

```python
# Import necessary libraries from Hugging Face Transformers
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the humanizer model and tokenizer through Hugging Face Transformers and return them
# for later use in the Streamlit app
def load_humanizer():
    model_name = "google/flan-t5-base"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
    return tokenizer, model

# Function to humanize text using the loaded model and tokenizer
def humanize_text(text, tokenizer, model, max_words):
    prompt = f"Rewrite this text in a more human, natural, and conversational way within {max_words} words:\n\n{text}"
    
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True)
    outputs = model.generate(**inputs, max_length=512)
    
    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return result
```

```python
# Utility functions for the AI Text Checker application

# Import necessary libraries
import matplotlib.pyplot as plt

# Function to plot a pie chart of AI vs Human scores
def plot_pie(ai_score, human_score):
    labels = ['AI Generated', 'Human Written']
    sizes = [ai_score, human_score]

    fig, ax = plt.subplots()
    ax.pie(
        sizes,
        labels=labels,
        autopct='%1.1f%%',
        startangle=90
    )
    ax.axis('equal')
    return fig

# Function to count words in a given text
def word_count(text):
    return len(text.split())
```

```python
import streamlit as st
from detector import load_detector, detect_ai
from humanizer import load_humanizer, humanize_text
from utils import plot_pie, word_count

st.set_page_config(page_title="AI Detector & Humanizer", layout="wide")

st.title("🧠 AI Content Detector + Humanizer")

# Sidebar controls
st.sidebar.header("⚙️ Settings")

max_input_words = st.sidebar.slider(
    "Maximum Input Words",
    min_value=50,
    max_value=1000,
    value=300
)

max_output_words = st.sidebar.slider(
    "Humanized Output Word Limit",
    min_value=50,
    max_value=500,
    value=150
)

# Load models
@st.cache_resource
def load_models():
    detector = load_detector()
    tokenizer, model = load_humanizer()
    return detector, tokenizer, model

detector, tokenizer, model = load_models()

# Input
text = st.text_area("📄 Paste your text here:")

if st.button("Analyze"):
    if not text.strip():
        st.warning("Please enter some text.")
    else:
        words = word_count(text)

        if words > max_input_words:
            st.error(f"Text exceeds {max_input_words} word limit.")
        else:
            # Detection
            ai_score, human_score = detect_ai(text, detector)

            st.subheader("📊 AI vs Human Score")

            col1, col2 = st.columns(2)

            with col1:
                st.metric("AI Generated (%)", ai_score)

            with col2:
                st.metric("Human Written (%)", human_score)

            # Pie chart
            fig = plot_pie(ai_score, human_score)
            st.pyplot(fig)

            # Humanized output
            st.subheader("✍️ Humanized Version")

            humanized = humanize_text(
                text,
                tokenizer,
                model,
                max_output_words
            )

            st.write(humanized)

            st.download_button(
                "📥 Download Humanized Text",
                humanized,
                file_name="humanized.txt"
            )
```
